```python
df = df.drop(columns=["Id"])

print("Id column removed.")
print(df.shape)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="quality")

plt.title("Distribution of Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Number of Wines")

plt.show()
```

```python
quality_counts = df["quality"].value_counts().sort_index()

print(quality_counts)
```

```python
quality_percentage = df["quality"].value_counts(normalize=True).sort_index() * 100

print(quality_percentage.round(2))
```

```python
df.hist(
    figsize=(16, 12),
    bins=20
)

plt.suptitle("Distribution of Wine Characteristics", fontsize=16)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="alcohol",
    bins=20,
    kde=True
)

plt.title("Distribution of Alcohol Content")
plt.xlabel("Alcohol (%)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="quality",
    y="alcohol"
)

plt.title("Alcohol Content by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol (%)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="alcohol",
    y="quality"
)

plt.title("Relationship Between Alcohol and Wine Quality")
plt.xlabel("Alcohol (%)")
plt.ylabel("Wine Quality")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="quality",
    y="volatile acidity"
)

plt.title("Volatile Acidity by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Volatile Acidity")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="quality",
    y="sulphates"
)

plt.title("Sulphates by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Sulphates")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="quality",
    y="pH"
)

plt.title("pH by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("pH")

plt.show()
```

```python
correlation = df.corr()

correlation
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix of Wine Characteristics")

plt.show()
```

```python
quality_corr = df.corr()["quality"].sort_values(ascending=False)

print(quality_corr)
```

```python
plt.figure(figsize=(8, 6))

quality_corr.drop("quality").sort_values().plot(
    kind="barh"
)

plt.title("Correlation of Wine Characteristics with Quality")
plt.xlabel("Correlation with Quality")
plt.ylabel("Feature")

plt.show()
```

```python
plt.figure(figsize=(16, 8))

sns.boxplot(data=df)

plt.title("Boxplot of Wine Characteristics")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["alcohol"])

plt.title("Alcohol Outliers")

plt.show()
```

```python
selected_features = [
    "alcohol",
    "volatile acidity",
    "sulphates",
    "citric acid",
    "quality"
]

sns.pairplot(
    df[selected_features],
    hue="quality"
)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

df.groupby("quality")["alcohol"].mean().plot(
    kind="bar"
)

plt.title("Average Alcohol Content by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Average Alcohol (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

df.groupby("quality")["volatile acidity"].mean().plot(
    kind="bar"
)

plt.title("Average Volatile Acidity by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Average Volatile Acidity")
plt.xticks(rotation=0)

plt.show()
```

```python
# Key Insights

## 1. Dataset Overview

The dataset contains 1,143 red wine observations and 12 analytical features after removing the ID column.

## 2. Data Quality

The dataset contains no missing values and no duplicate records.

## 3. Wine Quality

The quality scores are distributed across multiple quality levels, with some quality categories occurring more frequently than others.

## 4. Alcohol and Quality

The relationship between alcohol content and wine quality was analyzed using boxplots, scatter plots, and correlation analysis.

## 5. Volatile Acidity and Quality

Volatile acidity was analyzed to understand whether wines with different quality scores show different acidity distributions.

## 6. Sulphates and Quality

Sulphate levels were compared across wine quality groups.

## 7. Correlation Analysis

The correlation matrix was used to identify numerical variables that have stronger statistical relationships with wine quality.

## 8. Outliers

Boxplots were used to identify potential outliers in the chemical characteristics of the wine.
```
